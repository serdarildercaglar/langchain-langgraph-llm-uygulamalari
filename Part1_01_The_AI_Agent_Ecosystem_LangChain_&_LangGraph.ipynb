# The Agent Stack: Frameworks, Runtimes, and Harnesses

To build effective AI agents, it is crucial to choose the right level of abstraction. The ecosystem is divided into three distinct layers: **Frameworks** (Integration), **Runtimes** (Orchestration), and **Harnesses** (Opinionated Systems).

Understanding where these layers overlap and how they differ will ensure you build on the right foundation.

---

## 1. Agent Frameworks (LangChain)

**The "Integration & Abstraction" Layer**

Agent frameworks are designed to lower the barrier to entry. They provide the necessary abstractions to connect LLMs to the outside world without requiring deep knowledge of underlying infrastructure.

* **Role:** Acts as the "middleware" between your code and the LLM.
* **Key Abstractions:** Structured prompts, tool interfaces, document loaders, and basic agent loops.
* **Relationship:** LangChain is built *on top of* LangGraph. It uses LangGraph's power but hides the complexity behind simple, user-friendly interfaces.

**Competitors:** Vercel AI SDK, CrewAI, LlamaIndex, OpenAI Agents SDK.

### When to use LangChain

Choose the Framework approach when speed and simplicity are your priorities:

* **Fast Prototyping:** You want to go from zero to a working agent in minutes.
* **Standard Needs:** You are building standard applications (e.g., a chatbot that searches PDFs) and don't need highly custom logic.
* **Broad Connectivity:** You need to connect to many different model providers (OpenAI, Anthropic, Gemini) using a single, unified API.

---

## 2. Agent Runtimes (LangGraph)

**The "Orchestration & Infrastructure" Layer**

Agent runtimes provide the production-grade tooling required to keep agents running reliably. While frameworks focus on *building* the agent, runtimes focus on *running* the agent, specifically handling state, failure, and time.

* **Role:** Acts as the "operating system" for the agent.
* **Key Capabilities:**
* **Durable Execution:** If the agent crashes or the server restarts, the runtime ensures the agent picks up exactly where it left off.
* **Persistence:** It manages memory across threads and sessions automatically.
* **Human-in-the-loop:** It allows execution to pause, wait for a human click or approval, and then resume.
* **Streaming:** It handles the complex flow of token streaming in multi-step workflows.



**Competitors:** Temporal, Inngest.

### When to use LangGraph

Choose the Runtime approach when control and reliability are your priorities:

* **Production Deployment:** You are moving from a demo to a business-critical application.
* **Complex Workflows:** Your agent needs to loop, retry, self-correct, or branch into parallel processes.
* **Long-Running Tasks:** The agent's task might take hours or days (e.g., "Research this company and write a report"), requiring state to be saved reliably.

---

## 3. Agent Harnesses (Deep Agents SDK)

**The "Batteries-Included" Layer**

Agent harnesses are the newest evolution in the stack. They are highly opinionated frameworks built on top of runtimes (like LangGraph) that come pre-packaged with advanced tools and architectural patterns.

* **Role:** Acts as a "pre-fabricated agent architecture."
* **Key Capabilities:**
* **Built-in Planning:** The agent automatically creates to-do lists and tracks its own progress.
* **File System Access:** Includes tools for reading/writing files and managing storage backends.
* **Sub-Agents:** Native ability to spawn "worker" agents to delegate sub-tasks to.
* **Token Management:** Automated summarization of history to prevent context-window overflows.



**Competitors:** Claude Agent SDK, Manus.

### When to use Deep Agents SDK

Choose the Harness approach when you need advanced capabilities out-of-the-box:

* **Sophisticated Automation:** You are building an agent that needs to behave like a human employee (planning tasks, using files, delegating work).
* **Reduced Boilerplate:** You don't want to write the code for "memory management" or "planning logic" from scratch.
* **Complex Multi-Step Tasks:** The task is too vague for a simple chain (e.g., "Plan a marketing campaign for next month") and requires decomposition.

## Feature Comparison Matrix

The table below illustrates how deep you need to integrate depending on the tool you choose.

| Feature | **LangChain** (Framework) | **LangGraph** (Runtime) | **Deep Agents** (Harness) |
| --- | --- | --- | --- |
| **Primary Goal** | Ease of Use | Control & Durability | Advanced Capabilities |
| **Short-term Memory** | Basic abstraction | Configurable state memory | Ephemeral State Backend |
| **Long-term Memory** | Vector store integrations | Persistence layer | Native Long-term memory module |
| **Agent logic** | You define the loop | You define the graph/nodes | **Pre-built** (Planning/Delegation) |
| **Sub-agents** | Manual implementation | Subgraphs (High control) | **Native** (Auto-delegation) |
| **Human-in-the-loop** | Middleware/Callbacks | Native Interrupts | "Interrupt_on" config |
| **Streaming** | Standard streaming | Event-based streaming | Harness-managed streaming |

### Summary Recommendation

* **Start with LangChain** if you are learning or building simple apps.
* **Upgrade to LangGraph** if you need custom control, loops, or production reliability.
* **Adopt Deep Agents** if you need a high-level, "smart" agent that can plan and use files immediately without writing the low-level logic yourself.
